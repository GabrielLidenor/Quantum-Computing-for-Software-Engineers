# What is a Vector ?

- Calculus, Volume 3, Section 2.1 ("Vectors in the Plane"), OpenStax.

Magnitude refers to the length, size, or absolute numerical value of the vector. It is always a non-negative real number.

Direction indicates where the vector points in space, typically measured by an angle relative to a reference axis (such as the positive x-axis), as shown in the diagram below:

A vector can be both a line vector and a column vector.

```python
import numpy as np

line_vector = np.array([1,2])

column_vector = np.array([[1],
                         [2]])

print(f"This is my line_vector ---> {line_vector}")
print(f"This is my column_vector ---> {column_vector}")
```

### Scalar multiplication

Vectors can also be multiplied by single numbers; these single numbers are called **scalars**

*"The word scalar (which means scale-like) is appropriate because scalar multiplication multiplies a vector by a factor, thereby changing the scale of the vector."*
- Axler, S. (2024). Linear Algebra Done Right (4th ed., p. 11). Springer.

When we multiply our vectors by a scalar we "scale" the vector; by scaling we mean we change the vector's length preserving its directional axis (or reversing its direction if the scalar is negative).

```python
vector_a = np.array([[1],
                      [2]])
scalar = 7

scaled_vector = scalar * vector_a

print(f"My vector a is now ---> {scaled_vector}")
```

### Linear combinations

"An expression of the form c1v1 + c2v2 + ... ckvk is called a linear combination of the vectors v1, v2, ...vk."
- Larson, R. (2017). Elementary Linear Algebra (8th ed., p. 165). Cengage Learning.

Remember the scalar we did in the previous section ?
When doing Linear combinations we scale the vectors and sum them to form a new vector.

```python
vector_a = np.array([[3],
                      [2]])

vector_b = np.array([[5],
                     [9]])

# to perform a linear combination, we must have the scalar.
# suppose we are given the scalar of 3
scalar_1 = 3
scalar_2 = 2

scaled_vector_a = scalar_1 * vector_a
scaled_vector_b = scalar_2 * vector_b

# now we just need to sum the scaled_vectors to have a new vector v

vector_v = scaled_vector_a + scaled_vector_b

print(f"My linear combination is: --->{vector_v}")
```

The linear combination in the code above can be written using ket notation as:

`|v⟩ = 3|a⟩ + 2|b⟩ = |c⟩ + |d⟩`

Where `|c⟩` and `|d⟩` are the new column vectors produced after scaling `|a⟩` and `|b⟩` by `3`.

Summing `|c⟩` and `|d⟩` yields our final combined vector `|v⟩`

### What is a Qubit ?

"*A quantum bit, or qubit for short... we're going to describe qubits as mathematical objects with certain specific properties... Just as a classical bit has a state either `0` or `1`, a qubit has states `|0⟩` and `|1⟩`.*

*The difference between bits and qubits is that a qubit can be in a state other than `|0⟩` or `|1⟩`. It is also possible to form linear combinations of states, often called superpositions: `|Ψ⟩`= `α|0⟩` + `β|1⟩` ..."*

- Nielsen, M. A., & Chuang, I. L., Quantum Computation and Quantum Information, 10th Anniversary ed. (Cambridge University Press, 2010), Chapter 1 (Section 1.2: Quantum bits).


In summary, qubits are mathematical representations of physical quantum phenomena. Scientists represent these quantum states using column vectors.

While classical bits represent either 0 or 1, quantum computing uses standard basis state vectors to represent these states:

$$|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \quad |1\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$$


### What Makes Quantum Computing So Special ?

Unlike a classical bit that must be strictly `0` or `1`, a qubit can exist in a **superposition** of both states simultaneously.

Mathematically, a superposition is simply a **linear combination** of $|0\rangle$ and $|1\rangle$:

$$|\psi\rangle = \alpha |0\rangle + \beta |1\rangle = \alpha \begin{pmatrix} 1 \\ 0 \end{pmatrix} + \beta \begin{pmatrix} 0 \\ 1 \end{pmatrix} = \begin{pmatrix} \alpha \\ \beta \end{pmatrix}$$

Where $\alpha$ and $\beta$ are scalar values called **probability amplitudes**.

Key properties of these probability amplitudes include:Complex Values:
* Although we often use real numbers in basic Python examples for easier visualization, quantum probability amplitudes are generally complex numbers ($\alpha, \beta \in \mathbb{C}$).
* Normalization Constraint: To ensure the total probability of all possible measurement outcomes equals $1$ ($100\%$), any valid quantum state must satisfy the normalization condition:

$$\vert{}\alpha\vert{}^2 + \vert{}\beta\vert{}^2 = 1$$

When measured, the qubit will collapse to state $\vert{}0\rangle$ with probability $\vert{}\alpha\vert{}^2$, or to state $\vert{}1\rangle$ with probability $\vert{}\beta\vert{}^2$.

### Superposition in Python

*"The superposition of two or more states produces a new quantum state. Conversely, any state can be expanded as a linear combination of a set of basis states."*
- Dirac, P. A. M., *The Principles of Quantum Mechanics* (4th ed., Oxford University Press, 1958).

A qubit exists in this combined linear state until we measure it, at which point the superposition collapses into one of the definite basis states ($|0\rangle$ or $|1\rangle$).

```python
# Basis states (our ket vectors)
ket_0 = np.array([[1],
                  [0]])
ket_1 = np.array([[0],
                  [1]])

# Superposition is just a linear combination.
# alpha and beta are probability amplitudes (scalars)
alpha = 1 / np.sqrt(2)
beta = 1 / np.sqrt(2)

# State psi in superposition: |psi> = alpha|0> + beta|1>
# or as we saw in the previous section, a linear combination
ket_psi = alpha * ket_0 + beta * ket_1

print(f"Quantum State |psi>:\n{ket_psi}")
```
